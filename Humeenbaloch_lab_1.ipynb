```python

print("✈️ Travel Agent")

destination = input("Where do you want to travel? ")
budget = int(input("Enter your budget: "))

if budget >= 30000:
    print("You can travel to", destination, "with a comfortable budget!")
elif budget >= 15000:
    print("You can travel to", destination, "with a basic plan.")
else:
    print("Your budget is low. Try a cheaper destination.")
```
